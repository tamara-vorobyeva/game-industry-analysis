# Assumptions Made


**Data and Methodology Limitations**

During our research, we encountered the following data access constraints:

- **Lack of Financial Metrics and In-Game Revenue**: Market volume is estimated solely by the number of copies sold. This skews the overall picture, particularly for the free-to-play segment.

- **Varying Release Dates**: The dataset only tracks sales from the moment of a game's latest release. Consequently, older titles (e.g., released in 2000) accumulate more sales by the end of the target period (2013) than newer releases (e.g., from 2010).

- **Implication**: These factors may artificially inflate the performance of premium games and the North American market (which historically launched earlier), even though the industry is currently driven by the mobile sector, handheld consoles, and Asia.

**Handling Missing Data**

Since this study is marketing-oriented rather than strategic, the following approach was applied to handle minor data discrepancies:

- Rows with isolated missing values (less than 1%) will be removed.

- Missing values in columns used for calculations will be imputed with mean values.

Some numerical data (EU sales, JP sales, User Score) are recorded in string format instead of float64.

Below, we will convert these columns to the required data type.

## Presence of missing values in the data

```python
# Calculate the number of missing values in each column in absolute and relative terms.
empty_num = df.isna().sum()
empty_share = (empty_num * 100 / df.shape[0]).round(4) 
missing_data = pd.DataFrame({
    'Missing values (units)': empty_num,
    'Share missing values (%)': empty_share
})
display(missing_data)
```

The highest number of missing values is in the "critic_score", "user_score", and "rating" columns. These are likely systematic gaps due to historical or technical reasons:

* **Pre-internet era:** The game was released before the rise of review websites (early 2000s).
* **Niche games:** The game was too obscure to be reviewed or didn't meet the minimum user rating threshold.
* **ESRB fees:** The ESRB rating requires a fee. If a game was only released in Japan or Europe, developers had no reason to pay for it.
* **Technical errors:** Data could have been lost during scraping, collection, or uploading.

<div class="alert alert-info">
<h2> Missing Value Handling Notes <a class="tocSkip"> </h2>

Missing values were handled to enable correct data types and calculations in Python: <br>
    - <b>rating:</b> Replaced NaN with "unknown". <br>
    - <b>name, year_of_release, genre:</b> Dropped rows with missing essential data. <br>
    - <b>eu_sales, jp_sales:</b> Filled missing values with the mean grouped by year and platform. <br>
    - <b>year_of_release:</b> Converted data type to int64.
</div>

```python
# In the "rating" column containing string values, replace NaN with "Unknown".
df['rating'] = df['rating'].fillna('Unknown')
```

```python
# Drop rows with missing essential information (game name, release year, genre) in columns that will be required for calculations.
df = df.dropna(subset=['name', 'genre', 'year_of_release'])

# Change the data format in the "year_of_release" column to int64.
df['year_of_release'] = df['year_of_release'].astype('int64')
```

```python
# For eu_sales and jp_sales, fill missing values with the mean grouped by year of release and platform.
mean_eu_sales = df[(df['year_of_release'] >= 2000) & (df['year_of_release'] <= 2013)].groupby(['year_of_release', 'platform'])['eu_sales'].mean()
mean_jp_sales = df[(df['year_of_release'] >= 2000) & (df['year_of_release'] <= 2013)].groupby(['year_of_release', 'platform'])['jp_sales'].mean()
eu_sales_table = pd.DataFrame(mean_eu_sales).reset_index()
jp_sales_table = pd.DataFrame(mean_jp_sales).reset_index()

print(f'Average sales volume per video game in Europe by release year and platform (units):')
display(eu_sales_table.sort_values(by='year_of_release', ascending=False).head(3))
print(f'Average sales volume per video game in Japan by release year and platform (pcs.)')
display(jp_sales_table.sort_values(by='year_of_release', ascending=False).head(3))
```

<div class="alert alert-info">
<b> Missing values in the eu_sales and jp_sales columns were filled with the mean values grouped by release year and platform.<br>

Since filtering the dataframe by year was not yet required at this stage of the study, we applied it solely within the scope of this task (filling missing values) to make the results for the period of interest more illustrative.</b>

<div class="alert alert-info">
<h2> Interim Conclusion for the "rating" column <a class="tocSkip"> </h2>

As can be seen, it contains several values that do not belong to the ESRB rating system.<br>
 
In reality, there are six main ratings: "E" (Everyone), "E10+" (Everyone 10 and Up), "T" (Teen), "M" (Mature), "AO" (Adults Only), and "RP" (Rating pending). Until 2018, the "EC" (Early childhood) rating was also used.<br>
 
Since we might group by rating in the future, it is necessary to standardize these erroneous values. The proposed solution is as follows:<br>
- Replace "K-A" (kids-adults) values with "E" (Everyone)<br>
- Replace "EC" value with "E".</b> 

```python
# Checking the "rating" column value distribution to decide how to handle erroneous values.
count_rating = df['rating'].value_counts()
share = round(count_rating * 100 / df['rating'].count(), 2)
rating_overview_df = pd.DataFrame({
    'Кол-во оценок': count_rating,
    'Доля (%)': share
})
display(rating_overview_df.sort_index(ascending=True))
```

```python
# Replace wrongly written rating values:
# - "K-A" (kids-adults) value with "E" (Everyone)
# - "EC" value with "E"

df['rating'] = df['rating'].str.replace('K-A', 'E')
df['rating'] = df['rating'].str.replace('EC', 'E')
```

# Data Categorization
    
Let's categorize the data: We will group all games by their "user_score" into the following categories: "high score" (from 8 to 10 inclusive), "medium score" (from 3 up to but not including 8), and "low score" (from 0 up to but not including 3). 


```python
df_actual['user_score_group'] = pd.cut(df_actual['user_score'], bins=[-0.1, 3, 8, 11], labels=["low score", "middle score", "high score"], right=False)
# Display the first 10 rows for verification.
display(df_actual[['name', 'user_score', 'user_score_group']].head(10))
```

- Divide all games by critic scores ("critic_score") and distinguish the following categories: "high score" (from 80 to 101 inclusive), "medium score" (from 30 to 80, excluding the right boundary of the interval), and "low score" (from 0 to 30, excluding the right boundary of the interval).

```python
df_actual['critic_score_group']  = pd.cut(df_actual['critic_score'], bins=[-0.1, 30, 80, 101], labels=["low score", "middle score", "high score"], right=False)

# Display the first 10 rows for verification.
display(df_actual[['name', 'critic_score', 'critic_score_group']].head(10))
```

- After categorizing the data, let's verify the result: group the data by the selected categories and count the number of games in each category.

```python
# "Calculating dataframe distribution by user rating categories and creating an output table.
count_user_val = df_actual['user_score_group'].value_counts()
share_user_val = (df_actual['user_score_group'].value_counts(normalize=True) * 100).round(2)

grouped_user_table = pd.DataFrame({
    'Score count': count_user_val,
    'Share (%)': share_user_val
})

display(grouped_user_table.sort_index())
```

```python
# Calculating dataframe distribution by сritic rating categories and creating an output table.

count_critic_val = df_actual['critic_score_group'].value_counts()
share_critic_val = (df_actual['critic_score_group'].value_counts(normalize=True) * 100).round(2)
grouped_critic_table = pd.DataFrame({
    'Score count': count_critic_val,
    'Share (%)': share_critic_val
})

display(grouped_critic_table.sort_index())
```

<div class="alert alert-info">

# Recommendations for Further Analysis
To ensure statistical integrity, the following methodological adjustments are recommended:

- **Exclude Free-to-Play (F2P) from Core Sales**: Omit F2P titles from primary sales analyses. Because standard metrics miss in-game revenue (microtransactions, battle passes), including them artificially deflates average revenue and distorts financial data.

- **Segment the Market**: For comprehensive views, strictly separate the dataset into Premium (Buy-to-Play) and Free-to-Play (F2P). Analyzing these independently prevents data contamination and yields more accurate insights.

- **Standardize Genre Comparisons via Sentiment**: Evaluate genre popularity using the mean values of User and Critic Scores. This normalizes the data, reflecting actual acclaim independent of sales volume.
